```python
import pandas as pd
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A4
```

<h3>Olhar a base de dados/importar a base de dados</h3>

```python
tabela = pd.read_csv("horas_extras.csv")
print(tabela)
```

<h3>Criar 1 relatório</h3>

```python
for linha in tabela.index:
    nome = tabela.loc[linha, "Nome"]
    departamento = tabela.loc[linha, "Departamento"]
    horas_extras = tabela.loc[linha, "Horas Extras"]
    mes_ref = tabela.loc[linha, "Referência"]

    documento_pdf = canvas.Canvas(f"Relatório de Horas Extras {nome}.pdf", A4)

    # Título
    documento_pdf.setFont("Helvetica-Bold", 18)
    documento_pdf.drawString(100, 750, f"Relatório de Horas Extras {nome}")

    # Escrever o texto
    documento_pdf.setFont("Helvetica", 12)
    documento_pdf.drawString(100, 700, f"Nome: {nome}")
    documento_pdf.drawString(100, 670, f"Departamento: {departamento}")
    documento_pdf.drawString(100, 640, f"Quantidade de Horas Extras: {horas_extras}")
    documento_pdf.drawString(100, 610, f"Mês de Referência {mes_ref}")
    documento_pdf.drawString(100, 580, "Esse relatório foi gerado automaticamente com Python")

    # Salvar os relatórios
    documento_pdf.save()
```
